```python

from pathlib import Path
import pandas as pd

DATA_DIR = Path.cwd()

# قراءة قاموس الدول
dictionary_path = DATA_DIR / "country_dictionary_v1.csv"
dictionary = pd.read_csv(dictionary_path, dtype={"ISO3": "string"})

dictionary["ISO3"] = dictionary["ISO3"].str.strip().str.upper()

print("Dictionary shape:", dictionary.shape)
print("Duplicate ISO3 codes:", dictionary["ISO3"].duplicated().sum())
print("Missing ISO3 codes:", dictionary["ISO3"].isna().sum())

# قراءة ملفات المؤشرات الثلاثة
raw_data = {}

excel_files = sorted(DATA_DIR.glob("*.xls"))

for file in excel_files:
    df = pd.read_excel(file, header=3, engine="xlrd")

    # تنظيف أسماء الأعمدة والمسافات في الأكواد
    df.columns = [str(col).strip() for col in df.columns]

    if "Country Code" not in df.columns:
        raise ValueError(
            f"Country Code column not found in {file.name}. "
            f"Columns: {df.columns.tolist()}"
        )

    df["Country Code"] = (
        df["Country Code"]
        .astype("string")
        .str.strip()
        .str.upper()
    )

    # اكتشاف أعمدة السنوات
    year_columns = [
        col for col in df.columns
        if str(col).strip().replace(".0", "").isdigit()
        and 1900 <= int(float(str(col))) <= 2100
    ]

    # تحويل قيم السنوات إلى أرقام
    for col in year_columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")

    raw_data[file.name] = df

    print("\n" + "=" * 65)
    print("File:", file.name)
    print("Shape after fixing headers:", df.shape)
    print("Columns:", df.columns.tolist()[:8], "...")
    print("Year range:", min(map(lambda x: int(float(x)), year_columns)),
          "to", max(map(lambda x: int(float(x)), year_columns)))
    print("Country-code duplicates:", df["Country Code"].duplicated().sum())
    print("Missing country codes:", df["Country Code"].isna().sum())

    print("\nMissing values in identifier columns:")
    print(
        df[["Country Name", "Country Code", "Indicator Name", "Indicator Code"]]
        .isna()
        .sum()
    )

    print("\nMost missing year columns (%):")
    missing_pct = df[year_columns].isna().mean().mul(100).sort_values(
        ascending=False
    )
    print(missing_pct.head(10).round(2))

    print("\nIndicator names:", df["Indicator Name"].dropna().unique())
    print("Indicator codes:", df["Indicator Code"].dropna().unique())

# مقارنة الأكواد الموجودة في الملفات بالقاموس
dictionary_codes = set(dictionary["ISO3"].dropna())

all_source_codes = set()

for df in raw_data.values():
    all_source_codes.update(df["Country Code"].dropna().tolist())

unmatched_codes = sorted(all_source_codes - dictionary_codes)

print("\n" + "=" * 65)
print("Unique source codes across all files:", len(all_source_codes))
print("Codes absent from dictionary:", len(unmatched_codes))
print(unmatched_codes)
```

```python

# فحص جودة البيانات قبل التحويل والدمج
import pandas as pd

quality_results = []

for file_name, df in raw_data.items():

    year_columns = [
        col for col in df.columns
        if str(col).isdigit()
    ]

    values = df[year_columns]

    missing_count = int(values.isna().sum().sum())
    total_cells = int(values.size)
    non_missing_count = int(values.notna().sum().sum())

    print("\n" + "=" * 65)
    print("File:", file_name)
    print("Country/entity rows:", len(df))
    print("Year columns:", len(year_columns))
    print("Total year cells:", total_cells)
    print("Available values:", non_missing_count)
    print("Missing values:", missing_count)
    print("Missing percentage:", round(missing_count / total_cells * 100, 2), "%")

    # السنوات التي لا تحتوي على أي قيمة
    empty_years = values.columns[values.isna().all()].tolist()
    print("Completely empty years:", empty_years)

    # السنوات التي تحتوي على بيانات
    yearly_coverage = values.notna().sum()
    print("\nYears with the most available values:")
    print(yearly_coverage.sort_values(ascending=False).head(10))

    # فحص القيم السالبة
    negative_count = int((values < 0).sum().sum())
    print("\nNegative numeric values:", negative_count)

    # فحص النطاقات الخاصة بالمؤشرات
    if "01_" in file_name:
        print("GDP values below zero:", negative_count)

    elif "02_" in file_name:
        outside_range = ((values < 0) | (values > 100)).sum().sum()
        print("Literacy values outside 0-100:", int(outside_range))

    elif "03_" in file_name:
        below_zero = int((values < 0).sum().sum())
        print("Enrollment values below zero:", below_zero)
        print("Enrollment values above 100:",
              int((values > 100).sum().sum()))

    quality_results.append({
        "File": file_name,
        "Rows": len(df),
        "Years": len(year_columns),
        "Available_Values": non_missing_count,
        "Missing_Values": missing_count,
        "Missing_Pct": round(missing_count / total_cells * 100, 2),
        "Negative_Values": negative_count,
    })

quality_report = pd.DataFrame(quality_results)

print("\n\nFINAL QUALITY SUMMARY")
display(quality_report)
```

```python

from pathlib import Path
import pandas as pd

# ==========================================
# 1. إعداد الملفات والأسماء النهائية
# ==========================================

output_dir = Path.cwd()

dictionary_file = output_dir / "country_dictionary_v1.csv"

output_csv = output_dir / "socioeconomic_education_clean.csv"
output_log = output_dir / "cleaning_log.txt"
unmatched_csv = output_dir / "unmatched_codes_review.csv"

dictionary = pd.read_csv(
    dictionary_file,
    dtype={"ISO3": "string"}
)

dictionary["ISO3"] = (
    dictionary["ISO3"].str.strip().str.upper()
)

dictionary["Country"] = dictionary["Country"].str.strip()

if dictionary["ISO3"].isna().any():
    raise ValueError("Dictionary contains missing ISO3 codes.")

if dictionary["ISO3"].duplicated().any():
    raise ValueError("Dictionary contains duplicate ISO3 codes.")

# نحتفظ فقط بأعمدة القاموس المطلوبة في الناتج
country_lookup = dictionary[["ISO3", "Country"]].copy()

# ==========================================
# 2. تحديد أسماء المؤشرات
# ==========================================

indicator_mapping = {
    "01_": "GDP_Per_Capita_PPP_2021",
    "02_": "Adult_Literacy_Rate",
    "03_": "Secondary_Enrollment_Gross_Pct",
}

expected_columns = [
    "GDP_Per_Capita_PPP_2021",
    "Secondary_Enrollment_Gross_Pct",
    "Adult_Literacy_Rate",
]

dictionary_codes = set(country_lookup["ISO3"])

cleaned_long = {}
source_stats = []
log_lines = []

log_lines.append("DATA CLEANING LOG")
log_lines.append("=" * 60)
log_lines.append(f"Dictionary rows: {len(dictionary)}")
log_lines.append(
    f"Unique dictionary ISO3 codes: {dictionary['ISO3'].nunique()}"
)

# ==========================================
# 3. فحص الأكواد وتحويل كل ملف إلى Long Format
# ==========================================

for file_name, df_original in raw_data.items():

    df = df_original.copy()

    # تحديد اسم العمود النهائي من اسم الملف
    prefix = next(
        (p for p in indicator_mapping if file_name.startswith(p)),
        None
    )

    if prefix is None:
        raise ValueError(f"Unknown indicator file: {file_name}")

    indicator_column = indicator_mapping[prefix]

    year_columns = [
        col for col in df.columns
        if str(col).isdigit()
    ]

    if not year_columns:
        raise ValueError(f"No year columns found in {file_name}")

    df["Country Code"] = (
        df["Country Code"]
        .astype("string")
        .str.strip()
        .str.upper()
    )

    # التأكد من عدم تكرار كود الدولة في الملف
    if df["Country Code"].duplicated().any():
        raise ValueError(f"Duplicate country codes in {file_name}")

    # تسجيل الأكواد غير الموجودة في القاموس
    source_codes = set(df["Country Code"].dropna())
    unmatched = sorted(source_codes - dictionary_codes)

    log_lines.append("")
    log_lines.append(f"Source file: {file_name}")
    log_lines.append(f"Source rows: {len(df)}")
    log_lines.append(f"Year columns: {len(year_columns)}")
    log_lines.append(f"Unmatched codes: {len(unmatched)}")
    log_lines.append(f"Unmatched code list: {', '.join(unmatched)}")

    # تحويل قيم السنوات إلى أرقام
    for col in year_columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")

    # الأكواد غير المطابقة تُسجل للمراجعة، ولا تُدمج
    matched_df = df[df["Country Code"].isin(dictionary_codes)].copy()

    # الاحتفاظ بالحقول اللازمة فقط
    matched_df = matched_df[
        ["Country Code"] + year_columns
    ]

    # Wide -> Long
    long_df = matched_df.melt(
        id_vars=["Country Code"],
        value_vars=year_columns,
        var_name="Year",
        value_name=indicator_column
    )

    long_df = long_df.rename(columns={"Country Code": "ISO3"})

    long_df["ISO3"] = long_df["ISO3"].astype("string")
    long_df["Year"] = pd.to_numeric(
        long_df["Year"], errors="raise"
    ).astype("int64")

    # التحقق من المفتاح قبل الدمج
    if long_df.duplicated(["ISO3", "Year"]).any():
        raise ValueError(
            f"Duplicate ISO3 + Year keys in {file_name}"
        )

    # فحص نطاقات المؤشرات: إيقاف التنفيذ عند وجود قيم غير صالحة
    if indicator_column == "GDP_Per_Capita_PPP_2021":
        invalid = long_df[indicator_column] < 0

    elif indicator_column == "Adult_Literacy_Rate":
        invalid = (
            (long_df[indicator_column] < 0)
            | (long_df[indicator_column] > 100)
        )

    else:
        # Gross enrollment may exceed 100%, but cannot be negative.
        invalid = long_df[indicator_column] < 0

    invalid_count = int(invalid.fillna(False).sum())

    log_lines.append(
        f"Invalid values in {indicator_column}: {invalid_count}"
    )

    if invalid_count:
        raise ValueError(
            f"Invalid values found in {indicator_column}. "
            "Review them before continuing."
        )

    cleaned_long[indicator_column] = long_df

    source_stats.append({
        "Indicator": indicator_column,
        "Source_Rows": len(df),
        "Matched_Codes": matched_df["Country Code"].nunique(),
        "Long_Rows": len(long_df),
        "Available_Values": int(long_df[indicator_column].notna().sum()),
        "Missing_Values": int(long_df[indicator_column].isna().sum()),
    })

# ==========================================
# 4. حفظ قائمة الأكواد غير المطابقة للمراجعة
# ==========================================

all_source_codes = set()

for df in raw_data.values():
    all_source_codes.update(
        df["Country Code"].dropna().str.strip().str.upper()
    )

all_unmatched_codes = sorted(all_source_codes - dictionary_codes)

unmatched_review = pd.DataFrame({
    "Unmatched_Code": all_unmatched_codes
})

unmatched_review.to_csv(
    unmatched_csv,
    index=False,
    encoding="utf-8-sig"
)

log_lines.append("")
log_lines.append(
    f"Total unique unmatched codes excluded from final dataset: "
    f"{len(all_unmatched_codes)}"
)
log_lines.append(
    f"Review file: {unmatched_csv.name}"
)

# ==========================================
# 5. دمج المؤشرات باستخدام ISO3 + Year
# ==========================================

merged_df = None

for indicator_column in [
    "GDP_Per_Capita_PPP_2021",
    "Secondary_Enrollment_Gross_Pct",
    "Adult_Literacy_Rate",
]:

    current_df = cleaned_long[indicator_column]

    if merged_df is None:
        merged_df = current_df.copy()
    else:
        rows_before = len(merged_df)

        merged_df = merged_df.merge(
            current_df,
            on=["ISO3", "Year"],
            how="outer",
            validate="one_to_one"
        )

        log_lines.append(
            f"Merging {indicator_column}: "
            f"rows before={rows_before}, rows after={len(merged_df)}"
        )

# إضافة اسم الدولة من القاموس
merged_df = merged_df.merge(
    country_lookup,
    on="ISO3",
    how="left",
    validate="many_to_one"
)

if merged_df["Country"].isna().any():
    raise ValueError("Some final rows have no country name.")

# ==========================================
# 6. التحقق من البيانات النهائية
# ==========================================

key_columns = ["ISO3", "Year"]

if merged_df.duplicated(key_columns).any():
    raise ValueError("Duplicate ISO3 + Year keys after merge.")

# عدد الصفوف قبل حذف الصفوف التي كل مؤشراتها Missing
rows_before_empty_removal = len(merged_df)

# حذف الصفوف التي لا تحتوي على أي مؤشر متاح
merged_df = merged_df.dropna(
    subset=expected_columns,
    how="all"
).copy()

# ترتيب الأعمدة
final_columns = [
    "ISO3",
    "Country",
    "Year",
    "GDP_Per_Capita_PPP_2021",
    "Secondary_Enrollment_Gross_Pct",
    "Adult_Literacy_Rate",
]

merged_df = merged_df[final_columns].sort_values(
    ["ISO3", "Year"]
).reset_index(drop=True)

# إعادة فحص المفتاح بعد الحذف والترتيب
if merged_df.duplicated(["ISO3", "Year"]).any():
    raise ValueError("Final dataset contains duplicate keys.")

if merged_df[expected_columns].isna().all(axis=1).any():
    raise ValueError("Rows with all indicators missing remain.")

# ==========================================
# 7. تقرير التغطية النهائية
# ==========================================

print("SOURCE SUMMARY")
source_summary = pd.DataFrame(source_stats)
display(source_summary)

print("\nFINAL DATASET")
print("Rows:", len(merged_df))
print("Unique countries:", merged_df["ISO3"].nunique())
print("Year range:", merged_df["Year"].min(), "-", merged_df["Year"].max())
print("Duplicate ISO3 + Year:", merged_df.duplicated(["ISO3", "Year"]).sum())
print("Rows removed because all indicators were missing:",
      rows_before_empty_removal - len(merged_df))

coverage_rows = []

for col in expected_columns:
    available = int(merged_df[col].notna().sum())
    missing = int(merged_df[col].isna().sum())
    total = len(merged_df)

    coverage_rows.append({
        "Indicator": col,
        "Available_Values": available,
        "Missing_Values": missing,
        "Missing_Pct": round(missing / total * 100, 2) if total else 0,
        "Coverage_Pct": round(available / total * 100, 2) if total else 0,
    })

coverage_report = pd.DataFrame(coverage_rows)
display(coverage_report)

# ==========================================
# 8. حفظ الناتج وسجل التنظيف
# ==========================================

merged_df.to_csv(
    output_csv,
    index=False,
    encoding="utf-8-sig"
)

log_lines.extend([
    "",
    "FINAL DATASET SUMMARY",
    f"Final rows: {len(merged_df)}",
    f"Final unique countries: {merged_df['ISO3'].nunique()}",
    f"Year range: {merged_df['Year'].min()} - {merged_df['Year'].max()}",
    f"Rows before removing all-missing records: {rows_before_empty_removal}",
    f"Rows removed because all indicators were missing: "
    f"{rows_before_empty_removal - len(merged_df)}",
    f"Final duplicate ISO3 + Year keys: "
    f"{merged_df.duplicated(['ISO3', 'Year']).sum()}",
    "",
    "INDICATOR COVERAGE",
])

for _, row in coverage_report.iterrows():
    log_lines.append(
        f"{row['Indicator']}: "
        f"available={row['Available_Values']}, "
        f"missing={row['Missing_Values']}, "
        f"missing_pct={row['Missing_Pct']}%, "
        f"coverage_pct={row['Coverage_Pct']}%"
    )

log_lines.extend([
    "",
    "Missing values were not imputed.",
    "No interpolation, forward fill, or backward fill was applied.",
    "Gross secondary enrollment values above 100% were retained.",
    "Source codes absent from the country dictionary were excluded "
    "from the final dataset and recorded for review.",
])

output_log.write_text("\n".join(log_lines), encoding="utf-8")

print("\nSAVED FILES")
print(output_csv)
print(output_log)
print(unmatched_csv)

print("\nPreview of final dataset:")
display(merged_df.head(10))
```
